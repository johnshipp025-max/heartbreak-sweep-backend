## 3. Load Configuration and Secrets

## 5. Execute Unit and Integration Tests
Run the repository's test suite and inspect results to confirm backend functionality.

```python
import subprocess
result = subprocess.run(['npm', 'test'], capture_output=True, text=True)
print('Return code:', result.returncode)
print(result.stdout)
print(result.stderr)
```

## 6. Explore API Endpoints
Send requests to backend endpoints and display response data to validate the API behavior.

```python
import requests
base_url = 'http://127.0.0.1:3000'  # adjust if running on a different port
endpoints = ['/status', '/api/health', '/']
for endpoint in endpoints:
    try:
        response = requests.get(base_url + endpoint, timeout=5)
        print(endpoint, response.status_code, response.text[:200])
    except Exception as exc:
        print(endpoint, 'error', exc)
```

## 7. Analyze Data Models and Schemas
Inspect the data models, schemas, and serialization logic used by the backend services.

```python
import re
for path in sorted(repo_root.glob('**/*.js')):
    if 'node_modules' not in str(path):
        text = path.read_text(encoding='utf-8', errors='ignore')
        if 'mongoose' in text or 'zod' in text or 'schema' in text.lower():
            print('---', path)
            for line in text.splitlines():
                if re.search(r'\b(zod|schema|mongoose|Model|DataTypes|sequelize)\b', line, re.IGNORECASE):
                    print(line.strip())
```
